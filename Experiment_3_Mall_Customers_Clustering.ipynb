```python

import pandas as pd

url = "https://raw.githubusercontent.com/gakudo-ai/open-datasets/refs/heads/main/Mall_Customers.csv"
mall = pd.read_csv(url)

print(mall.head())
print("Dataset shape:", mall.shape)
```

```python

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans, AgglomerativeClustering, DBSCAN
from sklearn.metrics import silhouette_score

X = mall[['Annual Income (k$)', 'Spending Score (1-100)']]
X_scaled = StandardScaler().fit_transform(X)

models = {
    'K-Means': KMeans(n_clusters=5, random_state=42, n_init=10),
    'Agglomerative': AgglomerativeClustering(n_clusters=5),
    'DBSCAN': DBSCAN(eps=0.5, min_samples=5)
}

results = []

for name, model in models.items():
    labels = model.fit_predict(X_scaled)
    mask = labels != -1
    clusters = set(labels) - {-1}

    score = (silhouette_score(X_scaled[mask], labels[mask])
             if len(clusters) >= 2 and len(set(labels[mask])) >= 2
             else None)

    results.append({
        'Algorithm': name,
        'Clusters': len(clusters),
        'Silhouette Score': round(score, 3) if score is not None else 'N/A'
    })

result_table = pd.DataFrame(results)
print(result_table.to_string(index=False))
```

```python

# Save the dataset and clustering results
mall.to_csv('/content/Mall_Customers.csv', index=False)
result_table.to_csv('/content/Clustering_Results.csv', index=False)

print("Dataset and results saved successfully!")
```
